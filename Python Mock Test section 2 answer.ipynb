```python
s = "Hello123@#"
upper=""; lower=""; digit=""; special=""

for ch in s:
    if ch.isupper(): 
        upper+=ch
    elif ch.islower(): 
        lower+=ch
    elif ch.isdigit(): 
        digit+=ch
    else: 
        special+=ch

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Digits:", digit)
print("Special:", special)
```
