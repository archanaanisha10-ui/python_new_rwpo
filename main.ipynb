```python
# Student Details

# Get student name, age and course using input(). Convert age to int and display the details using print().

name = input("Enter your name: ")
age = int(input("Enter your age: "))
course = input("Enter your course: ")

print("Name:", name)
print("Age:", age)
print("Course:", course)
```

```python
# Simple Calculator

# Get two numbers from the user and display addition, subtraction, multiplication, division, floor division and modulus.

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
```
