```python
a = 1
while a <= 5:
    b = 1
    while b <= 5:
        print("*", end='')
        b += 1
    a += 1
    print()
```

```python
row = int(input("Enter a Row Number : "))
column = int(input("Enter a column Number : "))
symbol = input("Enter a symbol : ")

a = 1
while a <= row:
    b = 1
    while b <= column:
        print(symbol, end='')
        b += 1
    a += 1
    print()
```

```python
row = int(input("Enter a Row Number :"))
column = int(input("Enter a column Number :"))
symbol = input("Enter a symbol Number :")

for x in range(row):
    for y in range(column):
        print(symbol,end='')
    print()
```
